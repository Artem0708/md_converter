```python
from markitdown import MarkItDown
from pathlib import Path
import extract_msg
```

```python
#path = input()
```

```python
# input_folder = Path(path)
input_folder = Path("input")
output_folder = Path(path).parent / "output"
# output_folder = Path(f"{path.parent}/output")
output_folder.mkdir(parents=True, exist_ok=True)
```

```python
md = MarkItDown(enable_plugins=False) # Set to True to enable plugins
for item in input_folder.iterdir():
    try:
        
        if item.is_file() and item.stem[0] != ".":
            if item.suffix == ".msg":
                msg = extract_msg.Message(item)
                result = f"# **Subject:**{msg.subject} \n **Date:**{msg.date} \n **From:**{msg.sender} \n **To:**{msg.to} \n--- \n {msg.body}"
                msg.close()
            else:
                result = md.convert(item).text_content
            save_path = output_folder / f"{item.stem}.md"
            save_path.write_text(result)
            print(f"Конвертировал: {item.name} -> {save_path}")

    except Exception as e:
        print(f"❌ Ошибка с файлом {item.name}: {e}")
        
print("🏁 Все задачи выполнены.")        
```
